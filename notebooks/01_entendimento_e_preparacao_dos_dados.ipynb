# Entendimento e preparação dos dados

Este notebook apresenta a etapa inicial do projeto, dedicada ao entendimento e à preparação da base de clientes.

O objetivo é compreender a estrutura dos dados, revisar os tipos das variáveis, traduzir e padronizar os nomes das colunas e verificar possíveis problemas de qualidade, como valores ausentes, campos vazios e registros duplicados.

Também serão removidas variáveis sem utilidade analítica ou com risco de vazamento de dados.

Ao final, a base tratada será exportada para ser utilizada no notebook de análise exploratória.

## Importação das bibliotecas

## 1. Preparação do ambiente

Nesta etapa serão importadas as bibliotecas utilizadas para leitura, manipulação e tratamento dos dados.

```python
import numpy as np
import pandas as pd
```

## 2. Leitura dos dados

Nesta etapa, a base original será carregada a partir da pasta de dados brutos.

Uma cópia do conjunto de dados será criada para preservar o arquivo original e garantir que os tratamentos sejam realizados de forma segura.

```python
df_original = pd.read_csv('/content/NPS.csv')

df = df_original.copy()
```

```python
# Visualizando as primeiras linhas
df.head()
```

## 3. Entendimento inicial da base

Antes de realizar qualquer tratamento, é importante compreender a estrutura do conjunto de dados.

Nesta etapa serão analisados:

- a quantidade de registros e variáveis;
- os nomes das colunas;
- os tipos de dados;
- as estatísticas descritivas das variáveis numéricas.

Esse diagnóstico inicial ajuda a identificar possíveis inconsistências e orienta as próximas etapas de preparação dos dados.

```python
# Verificando o tamanho da Base
print(f'Quantidade de linhas: {df.shape[0]}')
print(f'Quantidade de colunas: {df.shape[1]}')
```

```python
# Listando todas as colunas
df.columns.tolist()
```

```python
# Verificando os tipos dos dados
df.info()
```

```python
# Observando as estatísticas gerais
df.describe()
```

## 4. Tradução e padronização das colunas

Os nomes das colunas serão traduzidos para português e padronizados para facilitar a leitura, a interpretação e a manipulação dos dados ao longo do projeto.

A tradução será realizada por meio de um dicionário de mapeamento, preservando a estrutura original da base e tornando o processo mais claro e reproduzível.

```python
# Fazendo a tradução das colunas para um melhor entendimento
df.columns = ['id_cliente', 'idade_cliente', 'regiao_cliente', 'tempo_cliente_meses', 'id_order', 'valor_pedido', 'quantidade_itens','valor_desconto', 'parcelas_pagamento',
              'tempo_entrega_dias', 'atraso_entrega_dias', 'valor_frete','tentativas_entrega', 'contatos_atendimento', 'tempo_resolucao_dias',
              'nps', 'recompra_30_dias', 'numero_reclamacoes', 'pontuacao_csat_interno']
```

```python
# Verificando os nomes atualizados
df.head()
```

## 5. Verificação da qualidade dos dados

Nesta etapa, serão verificados valores ausentes, registros duplicados e possíveis inconsistências nas variáveis.

Essa análise é importante para confirmar se a base está adequada para as próximas etapas do projeto e identificar a necessidade de tratamentos adicionais.

```python
# Valores nulos dos dados
print(df.isnull().sum())
```

```python
# Verificando os dados duplicados no dataset
print(f'Total de dados duplicados: {df.duplicated().sum()}')
```

```python
# Verificando os valores únicos dos Dados
for col in df.columns:
  print(f'Coluna: {col}')
  print(df[col].unique())
  print('-'*50)
```

#### Resultado

Não foram identificados valores ausentes nas variáveis da base.

Dessa forma, não será necessário aplicar técnicas de remoção ou preenchimento de dados nesta etapa.

## 6. Seleção inicial de variáveis

Nesta etapa, serão removidas colunas que não agregam valor analítico ou que podem comprometer a interpretação dos resultados.

As variáveis `id_cliente` e `id_order` serão removidas por representarem identificadores únicos. Esses campos são úteis para rastreamento dos registros, mas não descrevem características do cliente, do pedido ou da experiência.

Também serão removidas as variáveis `recompra_30_dias` e `pontuacao_csat_interno`, pois representam informações posteriores ao momento da avaliação do NPS. A utilização dessas variáveis poderia introduzir vazamento de dados, conhecido como *data leakage*.

Essa seleção contribui para manter a consistência temporal da análise e evita que os resultados sejam influenciados por informações que não estariam disponíveis no momento real da avaliação.

```python
colunas_remover = [
    "id_cliente",
    "id_order",
    "recompra_30_dias",
    "pontuacao_csat_interno"
]

df = df.drop(
    columns=colunas_remover
)
```

```python
# verificando as colunas
df.columns.tolist()
```

### Resultado da seleção

Foram removidas quatro variáveis da base:

- duas colunas identificadoras;
- duas colunas com informações posteriores à avaliação do cliente.

Após essa etapa, permaneceram apenas as variáveis relacionadas ao perfil do cliente, às características do pedido, à entrega, ao atendimento e ao NPS.

A base resultante está mais adequada para a análise exploratória e para a etapa posterior de modelagem.

## 7. Exportação da base tratada

Após as etapas de entendimento, verificação da qualidade e seleção das variáveis, a base final será exportada em formato CSV.

O arquivo será salvo como `clientes_tratados.csv` e utilizado no próximo notebook, dedicado à análise exploratória dos dados.

A base original permanece preservada, enquanto esta nova versão contém apenas as variáveis consideradas adequadas para as próximas etapas do projeto.

```python
caminho_saida = "/content/clientes_tratados.csv"

df.to_csv(
    caminho_saida,
    index=False,
    encoding="utf-8"
)

print(f"Base exportada para: {caminho_saida}")
print(f"Dimensão da base: {df.shape}")
```
